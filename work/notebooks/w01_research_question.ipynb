## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

**Decision I'm improving:** which pages an editor should review first for a possible content refresh, given limited review capacity.

**Unit of analysis (grain):** one page (one `content_id`). One row in, one score out.

**Output:** a ranked review queue — each page gets a score and short reason codes explaining why it surfaced.

**Who acts on it:** a content editor or SEO reviewer, who works down the top of the queue (e.g. the top 50) and decides per page whether to refresh, expand, protect, or leave it alone.

**Cost of a wrong call:**
- *False positive* — a page ranked high that was actually fine: wasted editor hours reviewing something that didn't need it.
- *False negative* — a genuinely declining page pushed down the list: a real problem left unfixed while attention goes elsewhere.

Because review time is the scarce resource, the metric that matches this decision is Precision@K (how many of the top-K the reviewer actually needed to see), not overall accuracy.
